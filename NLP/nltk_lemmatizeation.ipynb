```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
tokens = ['houses','running','dogs','better','cats','flying','geese','studies','wolves','children','feet']
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
nltk.download('averaged_perceptron_tagger')
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
for word in tokens:
    print(word,lemmatizer.lemmatize(word))
```

```python
token_with_pos = nltk.pos_tag(tokens)
token_with_pos
```

NNS-->part of speech

```python
def pos_tag(pos):
    if pos.startswith('J'):
        return 'a'
    elif pos.startswith('N'):
        return 'n'
    elif pos.startswith('VBZ'):
        return 'v'
    elif pos.startswith('R'):
        return 'r'
    else:
        return 'n'
```

```python
for token,pos in token_with_pos:
    print(token,lemmatizer.lemmatize(token,pos_tag(pos)))
```
