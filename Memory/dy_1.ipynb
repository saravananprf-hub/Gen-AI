```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
llm_model = ChatGoogleGenerativeAI(
                                   model= 'gemini-3.8-flash',
                                   api_key = os.environ['Gemini_Api'],
                                   temperature = 0.2
                                   
)
```

```python
prompt = ChatPromptTemplate.from_messages([
                                            ("system","you are a helpful assistant"),
                                            (MessagesPlaceholder(variable_name="history")),
                                            #to insert the message inside prompt template
                                            ("human","{input}")
])
```

```python
chain_1 = prompt | llm_model | StrOutputParser()
store={}
```

```python
def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()#python memory
    return store[session_id]
```

```python
chain_with_memory = RunnableWithMessageHistory(chain_1,
                                               get_session_history,
                                               input_messages_key="input",
                                               history_messages_key='history')

user1= chain_with_memory.invoke(
                                {"input":"My name is Alice."},
                                config={"configurable":{"session_id":'user1'}}
)
user2= chain_with_memory.invoke(
                                {"input":"My name is Alice."},
                                config={"configurable":{"session_id":'user1'}}
)
```
