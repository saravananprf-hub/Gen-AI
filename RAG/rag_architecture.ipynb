```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader
from langchain_huggingface import HuggingFaceEmbeddings 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.vectorstores import FAISS
import spacy
import re
import contractions
from textblob import TextBlob
import os
from langchain_ollama import ChatOllama
```

**1.Load the document(.txt)**

```python
data = TextLoader('RAG.txt').load()
data = data[0].page_content
print(data)
```

```python
import string
var = string.punctuation
# print(var) --> !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
data=re.sub(f'[{var}]',' ',data)
print(data)
data=re.sub(r'\s+',' ',data)
print(data)
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 200,
    chunk_overlap = 25,
    separators=['\n\n','.','\n',' ','']
)
chunks = splitter.create_documents([data])
# for chunk in chunks:
#     print(chunk.page_content)
chunks
```

```python
# chunks[0].metadata={'file_name':'data.txt'}
# chunks
for chunk in range(len(chunks)):
    chunks[chunk].metadata={'file_name':'data.txt'}
chunks
        
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name = 'sentence-transformers/all-miniLM-L6-V2' # Encoder only transforms only architecture
)
embedding_model
```

**Chunks into vectors**

```python
pip install faiss-cpu
```

```python
vectodb = FAISS.from_documents(
    documents = chunks,
    embedding = embedding_model
)
vectodb
```

```python
user_query = 'Explain rag architecture?'
r_chunks =vectodb.similarity_search(user_query)  # it uses ecludiean distance formula for similarity search
for chunk in r_chunks:
    print(str(chunk.page_content))
```

**Retraival part**

```python
def retrival(query : str,k):
    pass
    r_chunks =vectodb.similarity_search(query,k=k)
    retraival = [chunk.page_content for chunk in r_chunks]
    retrival_text = '\n'.join(retraival)
    return retrival_text
try:
    retrival('Explain rag architecture?',2)
except Exception as e:
    print(f'Error:{e}')
    
```

```python
def genarate_text(qry,ans):
    prompt = f''' 
    you're a helpful assistant and structure the given content.
    input : {qry}
    output : {ans}

    note:
    1)Don't add extra contenta just structure my output.
    2)If the query is not related to context just say 'query is not related to content'.
    output Format:
    input :{qry}
    output : structured output

    '''
    try:
        response = model_gemini.invoke(prompt)
        return response.content[0]['text']
    except:
        response = model_ollama.invoke(prompt)
        return response.content
retrival_t = retrival(user_query,2)
final_response = genarate_text(user_query,retrival_t)
print(final_response)
```
