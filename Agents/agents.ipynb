```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain.agents import create_agent
import os
```

```python
# while creating a tool we should mention doc str
```

```python
@tool #tool(get_word_count)
def get_word_count(text:str) ->str:
    """Count the number of words in the given text.""" #doc string
    return f"{len(text.split())} words"

@tool 
def reverse_text(text:str) -> str:
    """Reverse the given text. """
    return text[::-1]

tools =[get_word_count,reverse_text]
# LLM --------------------------
llm = ChatGoogleGenerativeAI(model='gemini-3.5-flash',api_key = os.environ['gemini_api'])
# Create Agent-------------------------------
agent = create_agent(llm,tools)
#-----Run Agent------------------
response = agent.invoke({
    "messages" : [
        ("user","How many words in 'Hello world this is LangChain'? ")
    ]
})
print(response['messages'][-1].content)
```

```python
print(response['messages'][-1].content[0]['text'])
```
