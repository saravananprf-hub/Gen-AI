```python
search = DuckDuckGoSearchRun()
```

```python
tools = [search]
```

```python
llm_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.2-flash',
    api_key = os.environ['Gemini_Api'],
    temperature = 0
    )
```

```python
Ollama= ChatOllama(
    #model='qwen2.5:1.5b'
    model='llama3.2'
)
```

```python
agent= create_agent(llm_model,tools)
```

```python
response = agent.invoke({
                         "messages" : [
                             ("user","In Canva Company,how many peoples are fired today")
                             ]
})
print(response['messages'][-1].content)
```
