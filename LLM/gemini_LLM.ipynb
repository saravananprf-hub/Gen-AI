```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
import os
```

```python
llm_model = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ['Gemini_Api'],
    temperature = 0
    
)
```

```python
response = llm_model.invoke("Explain Machine Learning?")
'''
llm_model.invoke("Explain Machine Learning?")
send the request to the gemini model
request => "Explain Machine Learning"?
respones => AIMessage
'''
```

```python
print(response.content[0]['text'])
```

```python
while True:
    user_input=input('Query:')
    response = llm_model.invoke(user_input)
    print("AI Resopnse : {response}")
    c=input("Type \nNo - exit \nAny Character - continue")
    if c.lower() =='no':
        break
```
