```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_community.document_loaders import PyPDFLoader
```

*Data Loading*

```python
rawdata = PyPDFLoader(r"C:\Users\Saravanan\OneDrive\resume\Saravanan_V_Data_Analyst_Resume-2.pdf").load()
resume = ""
for i in range(len(rawdata)):
    resume += rawdata[i].page_content
resume
```

*LLM*

```python
ollama = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature=0.2
)
```

```python
gemini = ChatGoogleGenerativeAI(
    api_key = os.environ['gemini_api'],
    model = 'gemini-3.5-flash',
    temperature = 0.2
)
```

*Prompt Template*

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('system',"Give the only ATS score out of 100 for the given resume.Don't give extra content"),
    ('human',"{resume_content},{job_description}")
])
```

*Parser*

```python
parser = StrOutputParser()
```

```python
language_chain = prompt_template | gemini | parser
response = language_chain.invoke({"resume_content" : resume, "job_description" : jd})
```
