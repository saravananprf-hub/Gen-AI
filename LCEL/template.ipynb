```python
prompt_template = ChatPromptTemplate.from_messages([
                                                    ("system","Detect Native from the given context and return only detected language"),
                                                    ("human","{prompt}")])
```

Step 2 => To create LLM model


```python
llm_model1 = ChatGoogleGenerativeAI(
    model = 'gemini-3.8-flash',
    api_key = os.environ['Gemini_Api'],
    temperature = 0
    
)
```

```python
llm_model2= ChatOllama(
    model='qwen2.5:1.5b',
    temperature=0.2
)
```

Step 3 => Parser

```python
parser = StrOutputParser()
```

```python
language_chain = prompt_template|llm_model1 | parser
response =language_chain.invoke("Machine Learning our easyna subject,ithala niraya algorithm iruku,Machine Learning na, computer-ku data kuduthu, adhulendhu patterns learn panna vechu, future-la nadakkuradha predict panna train panradhu.")
```
