```python
import os 
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.document_loaders import PyPDFLoader
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

Data Loading

```python
data = PyPDFLoader('Saravanan_V_Data_Analyst_Resume-2.pdf').load()
resume=''
for i in range(len(data)):
    resume+=data[i].page_content
resume
```

```python
print(data)
```

LLM_MODELS

```python
llm_model1 = ChatGoogleGenerativeAI(
    model = 'gemini-3.8-flash',
    api_key = os.environ['Gemini_Api'],
    temperature = 0
    
)
```

```python
ollama = ChatOllama(
                    model = 'gwen2.5:1.5b',
                    temperature = 0.2
                    
)
```

### LCEL

Prompt Template

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('system' , "You are an information extraction assistant. Extract only the skill block from given context"),
    ('human','{prompt}')
])
```

Parser

```python
parser = StrOutputParser()
```

```python
language_chain_1 = prompt_template | llm_model1 | parser
response_1 = language_chain_1.invoke({'prompt':resume})
```

```python
print(response_1)
```

```python
language_chain_2 = prompt_template | ollama | parser
response_2 = language_chain_2.invoke({'prompt',resume})
```
