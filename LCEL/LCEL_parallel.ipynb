```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.runnables import RunnableParallel
```

```python
rawdata = PyPDFLoader(r"C:\Users\Saravanan\OneDrive\resume\Saravanan_V_Data_Analyst_Resume-2.pdf").load()
resume = ""
for i in range(len(rawdata)):
    resume += rawdata[i].page_content
resume
```

```python
ollama_model = ChatOllama(
    model = 'qwen2.5:1.5b',
    temperature = 0.2

)
```

```python
llm_model = ChatGoogleGenerativeAI(
                                   model= 'gemini-3.5-flash',
                                   api_key = os.environ['Gemini_Api'],
                                   temperature = 0.2
                                   
)
```

```python
parser = StrOutputParser()
```

```python
skill_prompt = ChatPromptTemplate.from_messages([
    ("system","Extract the skills from resume."),
    ("human","resume :{resume}")
])
skill_chain = skill_prompt | llm_model | parser
```

*LCEL 2  - Summarize the about project/exp from the resume* 

```python
summerize_prompt = ChatPromptTemplate([
    ('system',"Summarize the experience from the resume"),
    ("human","resume : {resume}")
])

summarizze_chain = summerize_prompt | ollama_model | parser
```

*LCEL - Find the missing skills from the resume for the data science role*

```python
missing_skill_prompt = ChatPromptTemplate.from_messages([
    ('system',"For the Data Science Role find the missing skills from the resume"),
    ("human","resume : {resume}")
])

missing_skill_chain = missing_skill_prompt | ollama_model | parser
```

```python
final_chain = RunnableParallel(
    skills = skill_chain,
    summarize = summarizze_chain,
    missing_skills = missing_skill_chain
)
```

```python
respose = final_chain.invoke({"resume":resume})
```
